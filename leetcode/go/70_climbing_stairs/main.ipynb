```go
func climbStairs(n int) int {

    memo := make(map[int]int)

    var count func(n int, memo map[int]int) int

    

    count = func(n int, memo map[int]int) int {
        if n <= 2 {
            return n
        }

        if val, exists := memo[n]; exists {
            return val
        }

        result := count(n-1, memo) + count(n-2, memo)

        memo[n] = result

        return result
        
    }

    return count(n, memo)
    
}

func main() {

    result := climbStairs(5)
    fmt.Println(result)
}
```
