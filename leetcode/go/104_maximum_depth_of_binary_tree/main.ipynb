```go
func maxDepth(root *TreeNode) int {
    if root == nil {
		return 0
	}

	depth := 0
	queue := []*TreeNode{root}

	for len(queue) > 0 {

		treelevel := len(queue)
		fmt.Println("treelevel:lenofq", treelevel)

		fmt.Println("depthb4++", depth)
		depth++
		fmt.Println("depthaftr++", depth)

		for i := 0; i < treelevel; i++ {
		treenode := queue[0]
		queue = queue[1:]

		
		fmt.Println("queue: queue[1:]", queue)	
		if treenode.Left != nil {
			queue = append(queue, treenode.Left)
		}
		if treenode.Right != nil {
			queue = append(queue, treenode.Right)
			}
		}
	}

	return depth

}

type TreeNode struct {
	Val int
	Left *TreeNode
	Right *TreeNode
}

func buildTree(array []interface{}) *TreeNode {
	if len(array) == 0 || array[0] == nil {
		return nil
	}

	root := &TreeNode{Val: array[0].(int)}
	queue := []*TreeNode{root}

	i := 1

	for i < len(array) {
		current := queue[0]
		queue = queue[1:]

		if i < len(array) && array[i] != nil {
			left := &TreeNode{Val: array[i].(int)}
			current.Left = left
			queue = append(queue, left)
		}
		i++

		if i < len(array) && array[i] != nil {
			right := &TreeNode{Val: array[i].(int)}
			current.Right = right
			queue = append(queue, right)
		}
		i++
	}

	return root
}

func main (){

	array := []interface{}{3,9,20,nil,nil,15,7}
	// array := []interface{}{1,nil,2}
	

	tree := buildTree(array)

	result := maxDepth(tree)
	fmt.Println(result)



}
```
