```go
func moveZeroes(nums []int)  {

	// if len(nums) <= 1 || len(nums) > 10000 {
	// 	return
	// }
    
	ptr1 := 0
	ptr2 := 0

	for ptr1 < len(nums) && ptr2 < len(nums){
		if nums[ptr1] == 0 && nums[ptr2] != 0 {
			nums[ptr1] = nums[ptr2]
			ptr1++
			ptr2++

		} else {
			ptr2++
		}
	}
	
}

func main() {

	nums := []int{0,1,0,3,12}
	// nums := []int{0}

	fmt.Println(nums)
	moveZeroes(nums)
	fmt.Println(nums)

	

}
```
