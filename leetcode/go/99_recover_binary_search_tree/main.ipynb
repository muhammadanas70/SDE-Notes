```go
import ("fmt"
		"log")

/**
 * Definition for a binary tree node.
 * type TreeNode struct {
 *     Val int
 *     Left *TreeNode
 *     Right *TreeNode
 * }
 */

type TreeNode struct {
	Val int
	Left *TreeNode
	Right *TreeNode
}

func buildTree(arr []*int) *TreeNode {
	if len(arr) == 0 || arr[0] == nil {
		return nil
	}

	root := &TreeNode{
		Val: *arr[0],
	}

	queue := []*TreeNode{
		root,
	}

	i := 1

	for len(queue) > 0 && i < len(arr) {
		curr := queue[0]
		queue = queue[1:]

		if i < len(arr) && arr[i] != nil {
			curr.Left = &TreeNode{
				Val: *arr[i],
			}
			queue = append(queue, curr.Left)
		}
		i++

		if i < len(arr) && arr[i] != nil {
			curr.Right = &TreeNode{
				Val: *arr[i],
			}
			queue = append(queue, curr.Right)
		}
		i++
	}
	return root
}

func ptr(v int) *int {
	return &v
}


func recoverTree(root *TreeNode)  {

	

	var first_violation *TreeNode
	var second_violation *TreeNode
	var previous_value *TreeNode

	var inorder func(node *TreeNode)

	inorder = func(node *TreeNode) {
		if node == nil {
			return
		}

		inorder(node.Left)

		if previous_value != nil && previous_value.Val > node.Val {
			if first_violation == nil {
				first_violation = previous_value
			}
			second_violation = node
		}
		previous_value = node

		inorder(node.Right)
	}

	inorder(root)
    
	if first_violation != nil && second_violation != nil{
		first_violation.Val, second_violation.Val = second_violation.Val, first_violation.Val
	}
	
}

func main()  {

	var root = buildTree([]*int{ptr(1), ptr(3), nil, nil, ptr(2)})

	recoverTree(root)
	
}
```
