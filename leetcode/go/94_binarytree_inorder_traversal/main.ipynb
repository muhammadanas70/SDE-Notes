```go
import "fmt"

type Node struct{
    Val int
    Left *Node
    Right *Node
}

func insert(arr []int, i int) *Node {
    var root *Node
    n:= len(arr)

    if i < n {
        root = &Node{
            Val: arr[i],
        }

        root.Left = insert(arr, 2*i+1)

        root.Right = insert(arr, 2*i+2)
    }
    return root
}

func inorder(root *Node) []int{

    arr := []int{}

    if root != nil {
        arr = append(arr, inorder(root.Left)...)
        arr = append(arr, root.Val)
        arr = append(arr, inorder(root.Right)...)

    }

    return arr
}

func main() {
    nodes := []int{1,2,3,4,5,6}

    root := insert(nodes, 0)

    arr := inorder(root)
    fmt.Println(arr)
}
```
