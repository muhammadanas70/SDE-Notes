```go
import "fmt"

func removeDuplicates(nums []int) int {

	ptr1 := 0
	ptr2 := 1

	for ptr2 < len(nums) {
	if nums[ptr1] == nums[ptr2]{
		ptr2++
		if nums[ptr1] == nums[ptr2]{
			ptr1 = ptr2
			ptr2 = ptr1 + 1
		} else {
			ptr1 = ptr2
			ptr2++
		}

	} else {
		nums[ptr1] = nums[ptr2]
		ptr2++
	}
	}

	return len(nums)

}

func main() {

	nums := []int{0,0,1,1,1,1,2,3,3}

	result := removeDuplicates(nums)
	fmt.Println(result)


}

```
