```go
func stackString(s string) string {

	// convert string to rune
	runes := []rune(s)



	arr := []rune{}
	for _, char := range runes {
		if char != '#' {
			arr = append(arr, char)
		} else {
			if len(arr) > 0{ // what happend to my our code
				arr = arr[:len(arr) - 1]

			}
			
		}
	}
	return string(arr)

}

func backspaceCompare(s string, t string) bool {

	sRune := []rune(stackString(s))
	tRune := []rune(stackString(t))

	if len(sRune) != len(tRune) {
		return false
	}

	for index, char := range sRune {
		if char == tRune[index]{
			continue
		} else {
			return false
		}
	}

	return true
    
}

func main ()  {

	// s := "ab#c"
	// t := "ad#c"

	s := "ab##"
	t := "c#d#"

	// s := "a#c"
	// t := "b"

	result := backspaceCompare(s, t)
	fmt.Println(result)


}
```
