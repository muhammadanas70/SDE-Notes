```go
import "math"

func containsNearbyDuplicate(nums []int, k int) bool {

    maps := make(map[int]int)

    // if len(nums) <= 1 || len(nums) >= 100000 {
    //     return false
    // }

    // if k <= 0 || k >= 100000 {
    //     return false
    // }

    
    i := 0

    for len(nums) > i {

        // fmt.Println(nums[i])
        

        index_value, key := maps[nums[i]]
        if key {
            maps[nums[i]] = i
            
            if math.Abs(float64(i - index_value)) <= float64(k) {
                return true
            }
        } else {
           
            maps[nums[i]] = i
        }
        i++

    }
    
    
    return false
}

func main() {

    // nums := []int{1,2,3,1}
    // k := 3

    // nums := []int{1,2,3,1}
    // k := 2

    nums := []int{1,0,1,1}
    k := 1

    result := containsNearbyDuplicate(nums, k)
    _ = result
    fmt.Println(result)

}

// # Input: nums = [1,2,3,1], k = 3
// # Output: true

// # Input: nums = [1,0,1,1], k = 1
// # Output: true

// # Input: nums = [1,2,3,1,2,3], k = 2
// # Output: false
```
