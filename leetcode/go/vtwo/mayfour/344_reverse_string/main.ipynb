```go
func reverseString(s []byte) []byte {

	for index, value := range s {
		fmt.Printf("index: %d | value: %c\n", index, value)

		
	}

	length := len(s) - 1
	ptr := 0

	for ptr < length {

		s[ptr], s[length] = s[length], s[ptr]

		ptr++
		length--
	}

	return s
    
}

func main () {
	s := []byte{'h','e','l','l','o'}
	// s := []byte{"H","a","n","n","a","h"}

	result := reverseString(s)
	fmt.Printf("result : %c", result)
}

// Input: s = ["h","e","l","l","o"]
// Output: ["o","l","l","e","h"]

// Input: s = ["H","a","n","n","a","h"]
// Output: ["h","a","n","n","a","H"]
```
