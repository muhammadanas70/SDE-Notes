```go
func rotateString(s string, goal string) bool {
    
	if len(s) == 0 || len(goal) == 0 {
		return false
	}
	if len(s) != len(goal) {
		return false
	}

	break_loop := s

	for true {
	
	first_char := string(s[0])
	s = s[1:]
	s = s[0:] + first_char

	if s == goal {
		return true
	}
	if s == break_loop {
		break
	}
}
	return false
}

func main()  {

	// s := "abcde"
	// goal := "cdeab"

	s := "abcde"
	goal := "abced"

	result := rotateString(s, goal)
	
	fmt.Println(result)
	_ = result
}

// Input: s = "abcde", goal = "cdeab"
// Output: true
// Input: s = "abcde", goal = "abced"
// Output: false
```
