```go
import "fmt"

// i_mid i for index
// v_target v for value

func search(numbers []int , target int ) {
	lo_i, hi_i := 0, len(numbers)

	for lo_i <= hi_i {

		mid_i = lo_i + (hi_i - lo_i) / 2

		switch  {
		case numbers[mid_i] == target:
			return mid_i
		case numbers[mid_i] < target:
			lo_i = mid_i + 1
		default:
			hi_i = mid_i - 1
		}
	}

	return -1
}
```
