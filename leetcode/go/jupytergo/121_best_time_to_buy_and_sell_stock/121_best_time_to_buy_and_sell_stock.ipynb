```go
package main

import "fmt"

func maxProfit(prices []int) int {

	ptr_0 := 0;
	ptr_1 := 1;

	result := 0;

	// fmt.Println(ptr_0, ptr_1)

	for ptr_1 < len(prices) {
		// fmt.Println(ptr_1)
		if prices[ptr_0] < prices[ptr_1] {
			result = prices[ptr_1] - prices[ptr_0]
		} else {
			ptr_0 = ptr_1
			ptr_1++
		}
		ptr_1++
	}

	return result
    
}

func main() {
    // fmt.Println("Hello, World!")

	prices := []int{7,1,5,3,6,4}
	// prices := []int{7,6,4,3,1}
	// prices := []int{10,1,5,6,7,1}
	// prices := []int{10,8,7,5,2}
	result := maxProfit(prices)
	fmt.Println(result)
}
```
