```go
func reverseStr(s string, k int) string {
    runes := []rune(s)
	if len(runes) <= 1 {
		return s
	}

	fmt.Println(runes)

	if len(runes) < k {

	
		ptr1 := 0
		ptr2 := len(runes) - 1

		for ptr1 < ptr2 {
			runes[ptr1], runes[ptr2] = runes[ptr2], runes[ptr1]
			ptr1++
			ptr2--
		}

		return string(runes)
	}

	if len(runes) >= k {

		ptr1 := 0
		ptr2 := k - 1

		for {

			for ptr1 < ptr2 {
				runes[ptr1], runes[ptr2] = runes[ptr2], runes[ptr1]
				ptr1++
				ptr2--
			}

			if len(runes) < ptr1 && len(runes) <= ptr2 {

				ptr1 = 2 * k
				ptr2 = (2 * k) + k - 1

			} else {
				break
			}
			

		}

		return string(runes)
	}

	return s

}

func main() {
	// s := "abcdefg"
	// k := 2
// Output: "bacdfeg"

	s := "abcd"
	k := 2
// Output: "bacd"

	// s := "hello"
	// k := 6 

	result := reverseStr(s, k)
	_ = result
	fmt.Println(result)
}
```
