```go
func missingNumber(nums []int) int {

	// len_nums := len(nums)
	// fmt.Println(len_nums)
	sum := 0
	sum_arr := 0
	for _, values := range nums {
		// fmt.Println(values)
		sum = values + sum
		
	}

	fmt.Println(sum)

	for i := 1; i <= len(nums); i++ {
		sum_arr = i + sum_arr
	}

	fmt.Println(sum_arr)

	result := sum_arr - sum
	

	return result
    
}

func main(){
    nums := []int{9,6,4,2,3,5,7,0,1}
	result := missingNumber(nums)
	fmt.Println(result)
}
```
