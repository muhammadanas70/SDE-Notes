```go
type TreeNode struct {
	Val int
	Left *TreeNode
	Right *TreeNode
}

func isSymmetric(root *TreeNode) bool {

	if root == nil {
		return true
	}

	var checkTree func(left *TreeNode, right TreeNode) bool

	checkTree = func(left *TreeNode, right *TreeNode) bool {
		if left == nil && right == nil {
			return true
		}

		if left == nil || right == nil || left.Val != right.Val {
			return false
		}

		return checkTree(left.Left, right.Right) && checkTree(left.Right, right.Left)
	}

	return checkTree(root.left, root.right)
}

func buildTree(tree []interface{}) *TreeNode {
	if len(tree) == 0 ||tree[0] == nil {
		return nil
	}

	root := &TreeNode{Val: tree[0].(int)}
	queue := []*TreeNode{root}

	i := 1

	for i < len(tree) {
		current := queue[0]
		queue := queue[1:]

		if i < len(tree) &&tree[i] != nil {
			left := &TreeNode{Val: tree[i].(int)}
			current.Left = left
			queue = append(queue, left)
		}
		i++

		if i < len(tree) &&tree[i] != nil {
			right := &TreeNode{Val: tree[i].(int)}
			current.Right = right
			queue = append(queue, right)
		}
		i++
		
	}
	return root
}

func main () {
	tree := []interface{}{1,2,2,3,4,4,3}
	// tree := []interface{}{1,2,2,nil,3,nil,3}
	// tree := []interface{}{1, 2, 2, 3, 4, 4, 3, 5, 6, 7, 8, 8, 7, 6, 5}
	// tree := []interface{}{1,2,2,2,null,2}

	root := buildTree(tree)

	result := isSymmetric(root)

	fmt.Println(result)
}
```
